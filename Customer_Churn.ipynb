# Data Understanding

| **Column Name**       | **Description**                                                               
| --------------------- | ------------------------------------------------------------------------------ 
| **CustomerID**        | Unique identifier assigned to each customer.                                         
| **Age**               | Age of the customer in years.                                                                 
| **Gender**            | Gender of the customer.                                                                   
| **Tenure**            | Number of months the customer has stayed with the company.                           
| **Usage Frequency**   | Frequency of service usage by the customer (e.g., average monthly usage rate).        
| **Support Calls**     | Number of calls made by the customer to the support center.                                  
| **Payment Delay**     | Number of times or days the customer delayed payments.                                        
| **Subscription Type** | Type of subscription plan the customer is enrolled in.                                   
| **Contract Length**   | Duration of the customer’s contract (in months or categorized terms).           
| **Total Spend**       | Total amount spent by the customer on the company’s services.                      
| **Last Interaction**  | Number of days since the customer last interacted with the company.                  
| **Churn**             | Target variable — indicates whether the customer has left (1) or stayed (0).

# Machine Learning Pipeline

### Split Data into Input Feature & Target Column

```python
x = df.drop('Churn', axis= 1)
y = df['Churn']
```

### Numerical Pipeline

```python
num_cols = x.select_dtypes(include='number').columns
num_cols
```

```python
from sklearn.preprocessing import RobustScaler, StandardScaler, MinMaxScaler
from sklearn.pipeline import Pipeline

scaler = StandardScaler()

numerical_scaler_pipeline = Pipeline(steps= [('Standard Scaler', scaler)])
numerical_scaler_pipeline
```

### Categorical Pipeline

```python
catg_cols = x.select_dtypes(include='O').columns
catg_cols
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

ohe = OneHotEncoder(drop= 'first', sparse_output= False)

ohe_pipeline = Pipeline(steps= [('Ohe Encoder', ohe)])
ohe_pipeline
```

```python
# from sklearn.preprocessing import OrdinalEncoder

# ord = OrdinalEncoder(categories= [ ['Basic', 'Standard', 'Premium'] ])

# ord_pipeline = Pipeline(steps= [ ('Ordinal Encoder', ord) ])
# ord_pipeline
```

```python
# Check imbalance
round(df['Churn'].value_counts(normalize= True) * 100, 2)
```

### Assign each column to the corresponding pipeline

```python
from sklearn.compose import ColumnTransformer

preprocessing = ColumnTransformer(transformers= [('Numerical Scaler Pipeline', numerical_scaler_pipeline, num_cols),
                                                 ('OneHot Encoder Pipeline', ohe_pipeline, catg_cols)
                                                ]
                                 )
preprocessing
```

## Select Best Model

```python
from sklearn.model_selection import cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from sklearn.svm import SVC


models = [
    ('Logistic Regression', LogisticRegression(random_state= 42, n_jobs= -1)),
    ('Random Forest', RandomForestClassifier(random_state= 42, n_jobs= -1)),
    ('KNN', KNeighborsClassifier(n_jobs= -1)),
    ('Decision Tree', DecisionTreeClassifier(random_state= 42)),
    ('CatBoost', CatBoostClassifier()),
    ('Gaussian NB', GaussianNB()),
    ('LightGBM', LGBMClassifier()),
    ('XGBoost', XGBClassifier()),
]

for model in models:

    model_pipeline = Pipeline(steps= [ ('Preprocessing', preprocessing),
                                       ('Model', model[1])])
    
    results = cross_validate(model_pipeline, x, y, scoring= 'f1', return_train_score= True, n_jobs= -1)

    print(model[0])
    print('Train F1 Score :', round(results['train_score'].mean() * 100, 2))
    print('Test F1 Score :', round(results['test_score'].mean() * 100, 2))
    print('AVG Fit Time :', round(results['fit_time'].mean(),2))
    print('-' * 100)
```

## Hyperparameter Tuning

```python
model_pipeline = Pipeline(steps= [ ('Preprocessing', preprocessing),
                                       ('Model', KNeighborsClassifier(n_jobs= -1))])
```

```python
from sklearn.model_selection import GridSearchCV

parameters = {'Model__n_neighbors' : [13, 15, 17]}

result = GridSearchCV(model_pipeline, param_grid= parameters, cv= 5, scoring= 'f1', n_jobs= -1, return_train_score= True)

result.fit(x, y)
```

```python
result.best_params_
```

```python
round(result.best_score_ *100, 2)
```
